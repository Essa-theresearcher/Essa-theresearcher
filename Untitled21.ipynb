```python
# ============================
# DIABETIC RETINOPATHY DETECTION - COMPLETE OPTIMIZED PIPELINE
# Mount → Unzip → Load → Train with Focal Loss → Threshold Tuning → Evaluation
# ============================

import os, shutil, numpy as np, hashlib
import tensorflow as tf
from sklearn.metrics import confusion_matrix, classification_report, recall_score
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
import seaborn as sns

# ============================
# STEP 0: GPU CHECK
# ============================
print("TensorFlow:", tf.__version__)
gpus = tf.config.list_physical_devices("GPU")
print("GPUs:", gpus)
if not gpus:
    print("\n⚠️ WARNING: No GPU detected. In Colab: Runtime → Change runtime type → GPU\n")

# ============================
# STEP 1: MOUNT GOOGLE DRIVE
# ============================
from google.colab import drive
drive.mount("/content/drive")

# ============================
# STEP 2: DEFINE PATHS
# ============================
ZIP_PATH = "/content/drive/MyDrive/diabetic-retinopathy-detection.zip"
EXTRACT_BASE = "/content/dr_dataset"
PROJECT_ROOT = os.path.join(EXTRACT_BASE, "diabetic-retinopathy-detection")
DATA_ROOT = os.path.join(PROJECT_ROOT, "augmented_resized_binary")
TRAIN_DIR = os.path.join(DATA_ROOT, "train")
VAL_DIR = os.path.join(DATA_ROOT, "val")
TEST_DIR = os.path.join(DATA_ROOT, "test")

# ============================
# STEP 3: UNZIP DATASET
# ============================
if os.path.exists(EXTRACT_BASE):
    shutil.rmtree(EXTRACT_BASE)
os.makedirs(EXTRACT_BASE, exist_ok=True)

print("\n📦 Unzipping dataset from Drive...")
os.system(f'unzip -o -q "{ZIP_PATH}" -d "{EXTRACT_BASE}"')

# Remove Mac metadata
mac_junk = os.path.join(EXTRACT_BASE, "__MACOSX")
if os.path.exists(mac_junk):
    shutil.rmtree(mac_junk)

print("✅ Project root exists:", os.path.exists(PROJECT_ROOT))
print("✅ Dataset root exists:", os.path.exists(DATA_ROOT))
print("Train classes:", os.listdir(TRAIN_DIR))
print("Val classes:", os.listdir(VAL_DIR))
print("Test classes:", os.listdir(TEST_DIR))

# ============================
# STEP 4: DATA LEAKAGE CHECK (CRITICAL)
# ============================
print("\n🔍 Checking for data leakage...")

def hash_sample(folder, n=1000):
    paths = []
    for root, _, files in os.walk(folder):
        for f in files:
            if f.lower().endswith(('.jpg', '.jpeg', '.png')):
                paths.append(os.path.join(root, f))
    if len(paths) == 0:
        return set()
    sample = np.random.choice(paths, size=min(n, len(paths)), replace=False)
    hashes = set()
    for p in sample:
        with open(p, 'rb') as fp:
            hashes.add(hashlib.md5(fp.read()).hexdigest())
    return hashes

train_h = hash_sample(TRAIN_DIR, 2000)
val_h = hash_sample(VAL_DIR, 2000)
test_h = hash_sample(TEST_DIR, 2000)

leak_train_val = len(train_h & val_h)
leak_train_test = len(train_h & test_h)
leak_val_test = len(val_h & test_h)

print(f"Train ∩ Val: {leak_train_val}")
print(f"Train ∩ Test: {leak_train_test}")
print(f"Val ∩ Test: {leak_val_test}")

if leak_train_test > 0 or leak_val_test > 0:
    print("⚠️ WARNING: Data leakage detected! Metrics may be inflated.")
else:
    print("✅ No leakage detected.")

# ============================
# STEP 5: SPEED OPTIMIZATIONS
# ============================
from tensorflow.keras import mixed_precision
mixed_precision.set_global_policy("mixed_float16")
print("\n⚡ Mixed precision enabled:", mixed_precision.global_policy())

BATCH = 64
IMG_SIZE = (224, 224)
SEED = 42

# ============================
# STEP 6: BUILD DATASETS
# ============================
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR, image_size=IMG_SIZE, batch_size=BATCH,
    label_mode="binary", seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR, image_size=IMG_SIZE, batch_size=BATCH,
    label_mode="binary", shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR, image_size=IMG_SIZE, batch_size=BATCH,
    label_mode="binary", shuffle=False
)

class_names = train_ds.class_names
print("Class names:", class_names)  # Expected: ['affected', 'healthy']

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

# Extract labels for class weighting
train_labels = np.concatenate([y.numpy() for _, y in train_ds], axis=0).ravel()
print("Train samples:", len(train_labels))
print("Class distribution:", np.bincount(train_labels.astype(int)))

# ============================
# STEP 7: FOCAL LOSS (BETTER THAN CLASS WEIGHTING)
# ============================
def focal_loss(gamma=2.0, alpha=0.75):
    """
    Focal Loss for binary classification.
    Focuses training on hard-to-classify examples.
    alpha: weight for positive class (affected=0 in our case, but applied after sigmoid)
    gamma: focusing parameter (higher = focus more on hard examples)
    """
    def loss_fn(y_true, y_pred):
        epsilon = tf.keras.backend.epsilon()
        y_pred = tf.clip_by_value(y_pred, epsilon, 1.0 - epsilon)

        # Binary focal loss
        pt = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
        focal_weight = tf.pow(1.0 - pt, gamma)
        loss = -alpha * focal_weight * tf.math.log(pt)
        return tf.reduce_mean(loss)
    return loss_fn

# ============================
# STEP 8: BUILD MODEL (EfficientNetB0)
# ============================
base = tf.keras.applications.EfficientNetB0(
    include_top=False, weights="imagenet", input_shape=(224, 224, 3)
)
base.trainable = False

inputs = tf.keras.Input(shape=(224, 224, 3))
x = tf.keras.applications.efficientnet.preprocess_input(inputs)
x = base(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.3)(x)
outputs = tf.keras.layers.Dense(1, activation="sigmoid", dtype="float32")(x)

model = tf.keras.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),
    loss=focal_loss(gamma=2.0, alpha=0.75),
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc"),
    ]
)

print("\n📊 Model Summary:")
model.summary()

# ============================
# STEP 9: CALLBACKS
# ============================
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_auc", mode="max", patience=3, restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_auc", mode="max", factor=0.5, patience=2
    ),
    tf.keras.callbacks.ModelCheckpoint(
        "dr_focal_phase1.keras", monitor="val_auc", mode="max", save_best_only=True
    )
]

# ============================
# STEP 10: PHASE 1 TRAINING (FROZEN BACKBONE)
# ============================
print("\n🔥 Starting Phase 1 Training (Frozen Backbone)...")
EPOCHS_PHASE1 = 10

history1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_PHASE1,
    callbacks=callbacks
)

# ============================
# STEP 11: PHASE 2 FINE-TUNING (UNFREEZE TOP LAYERS)
# ============================
print("\n🔥 Starting Phase 2 Fine-Tuning...")
base.trainable = True
fine_tune_at = int(len(base.layers) * 0.7)
for layer in base.layers[:fine_tune_at]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),
    loss=focal_loss(gamma=2.0, alpha=0.75),
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall"),
        tf.keras.metrics.AUC(name="auc"),
    ]
)

callbacks_ft = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_auc", mode="max", patience=3, restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_auc", mode="max", factor=0.5, patience=2
    ),
    tf.keras.callbacks.ModelCheckpoint(
        "dr_focal_finetuned.keras", monitor="val_auc", mode="max", save_best_only=True
    )
]

EPOCHS_FT = 10
history2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_FT,
    callbacks=callbacks_ft
)

# ============================
# STEP 12: TEST EVALUATION
# ============================
print("\n📈 Test Evaluation:")
test_metrics = model.evaluate(test_ds, verbose=0)
print(dict(zip(model.metrics_names, test_metrics)))

# ============================
# STEP 13: GET PREDICTIONS
# ============================
y_true = np.concatenate([y.numpy().astype(int) for _, y in test_ds], axis=0).ravel()
y_prob = model.predict(test_ds, verbose=0).ravel()

print("\n🎯 Class encoding:")
print(f"class_names: {class_names}")
print(f"0 = {class_names[0]}, 1 = {class_names[1]}")
print(f"y_prob = P(class=1) = P({class_names[1]})")

# ============================
# STEP 14: THRESHOLD TUNING FOR 90%+ RECALL
# ============================
print("\n🔧 Finding optimal threshold for ≥90% recall (affected)...")

def recall_at_threshold(t):
    y_pred = (y_prob >= t).astype(int)
    cm = confusion_matrix(y_true, y_pred)
    tp = cm[0, 0]
    fn = cm[0, 1]
    fp = cm[1, 0]
    recall_aff = tp / (tp + fn + 1e-9)
    return recall_aff, fp, cm

target_recall = 0.90
best_threshold = None

for t in np.linspace(0.5, 0.99, 100):
    r_aff, fp, cm = recall_at_threshold(t)
    if r_aff >= target_recall:
        best_threshold = t
        print(f"\n✅ Threshold = {t:.3f}")
        print(f"   Recall (affected) = {r_aff:.3f}")
        print(f"   False Positives = {fp}")
        print(f"   Confusion Matrix:\n{cm}")
        break

if best_threshold is None:
    print("⚠️ Could not reach 90% recall. Showing best available:")
    results = [(t, *recall_at_threshold(t)) for t in np.linspace(0.5, 0.99, 50)]
    results.sort(key=lambda x: x[1], reverse=True)
    best_threshold = results[0][0]
    print(f"Best threshold = {best_threshold:.3f}, recall = {results[0][1]:.3f}")

# ============================
# STEP 15: FINAL EVALUATION AT OPTIMAL THRESHOLD
# ============================
y_pred_optimal = (y_prob >= best_threshold).astype(int)

print("\n" + "="*60)
print(f"FINAL RESULTS AT THRESHOLD = {best_threshold:.3f}")
print("="*60)

cm_final = confusion_matrix(y_true, y_pred_optimal)
print("\nConfusion Matrix:")
print(cm_final)

print("\nClassification Report:")
print(classification_report(y_true, y_pred_optimal, target_names=class_names))

# ============================
# STEP 16: VISUALIZATION
# ============================
plt.figure(figsize=(8, 6))
sns.heatmap(cm_final, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title(f"Confusion Matrix (Threshold={best_threshold:.3f})")
plt.tight_layout()
plt.savefig("confusion_matrix_optimal.png", dpi=150)
print("\n💾 Confusion matrix saved as 'confusion_matrix_optimal.png'")

# ============================
# STEP 17: SAVE TO GOOGLE DRIVE
# ============================
SAVE_DIR = "/content/drive/MyDrive/DR_Models_Optimized"
os.makedirs(SAVE_DIR, exist_ok=True)

# Save model
model_path = os.path.join(SAVE_DIR, "dr_focal_final.keras")
model.save(model_path)
print(f"\n💾 Model saved to: {model_path}")

# Save threshold
threshold_path = os.path.join(SAVE_DIR, "optimal_threshold.txt")
with open(threshold_path, 'w') as f:
    f.write(f"optimal_threshold={best_threshold:.4f}\n")
    f.write(f"target_recall_affected=0.90\n")
print(f"💾 Threshold saved to: {threshold_path}")

# Copy checkpoints
for ckpt in ["dr_focal_phase1.keras", "dr_focal_finetuned.keras"]:
    if os.path.exists(ckpt):
        shutil.copy(ckpt, os.path.join(SAVE_DIR, ckpt))
        print(f"💾 Checkpoint saved: {ckpt}")

# Save confusion matrix
shutil.copy("confusion_matrix_optimal.png", os.path.join(SAVE_DIR, "confusion_matrix_optimal.png"))

print("\n" + "="*60)
print("✅ PIPELINE COMPLETE!")
print("="*60)
print(f"\nAll files saved to: {SAVE_DIR}")
print("\n📊 Summary:")
print(f"   - Best Val AUC: {max(history2.history['val_auc']):.4f}")
print(f"   - Optimal Threshold: {best_threshold:.4f}")
print(f"   - Test Accuracy: {test_metrics[1]:.4f}")
print(f"   - Recall (affected): Check classification report above")
```
